```python
import pandas as pd
```

```python
data = {
    "video_id": [
        "V01", "V02", "V03", "V04", "V05",
        "V06", "V07", "V08", "V09", "V10"
    ],
    "sign": [
        "HELLO", "HELLO", "HELLO", "HELP", "HELP",
        "HELP", "YES", "YES", "NO", "NO"
    ],
    "signer_id": [
        "S01", "S01", "S02", "S01", "S02",
        "S03", "S02", "S03", "S03", "S03"
    ],
    "environment": [
        "indoor", "indoor", "outdoor", "outdoor", "outdoor",
        "outdoor", "indoor", "indoor", "indoor", "outdoor"
    ],
    "quality_score": [5, 4, 5, 2, 3, 2, 5, 4, 5, 4]
}
```

```python
df = pd.DataFrame(data)
```

```python
# Task 1 — Basic aggregation
videos_per_sign = df["sign"].value_counts()
videos_per_signer = df["signer_id"].value_counts()
videos_per_environment = df["environment"].value_counts()
```

```python
# Task 2 — Average quality
average_quality_by_sign = df.groupby("sign")["quality_score"].mean()
print(average_quality_by_sign)
```

```python
# Task 3 — Two-variable analysis
videos_by_sign_environment = df.groupby(["sign", "environment"]).size()
print(videos_by_sign_environment)
```
