```python
import pandas as pd
```

```python
videos = pd.DataFrame({
    "video_id": [
        "V01", "V02", "V03", "V04", "V05",
        "V06", "V07", "V08", "V09", "V10"
    ],
    "sign": [
        "HELLO", "HELLO", "HELLO", "HELP", "HELP",
        "HELP", "YES", "YES", "NO", "NO"
    ],
    "signer_id": [
        "S01", "S01", "S02", "S01", "S02",
        "S03", "S02", "S03", "S03", "S03"
    ]
})

metadata = pd.DataFrame({
    "video_id": [
        "V01", "V02", "V03", "V04", "V05",
        "V06", "V07", "V08", "V09"
    ],
    "environment": [
        "indoor", "indoor", "outdoor", "outdoor", "outdoor",
        "outdoor", "indoor", "indoor", "indoor"
    ],
    "quality_score": [5, 4, 5, 2, 3, 2, 5, 4, 5]
})
```

```python
#Task 1 — Inner Merge
  #merged_inner Code:
merged_inner = pd.merge(videos, metadata, on="video_id", how="inner")
#Number of rows: 9
#Which video is missing: V10
#Why it is missing: An inner merge keeps only records where the video_id key is present in both DataFrames. Since V10 exists in videos but is absent from metadata, Pandas drops it.
```

```python
#Task 2 — Left Merge
  #merged_left Code:
merged_left = pd.merge(videos, metadata, on="video_id", how="left")
print(merged_left)
```

```python
#Task 3 — Detect the Missing Metadata
  # Filtering Code:
missing_metadata = merged_left[merged_left["environment"].isna()]
print(missing_metadata)
```

```python
#Task 4 — Research Summary
  #How many videos have environment information?
merged_left["environment"].count()
#How many videos have missing quality information?
merged_left["quality_score"].isna().sum()
```
