```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
data = {
    "video_id": ["V01", "V02", "V03", "V04", "V05",
                 "V06", "V07", "V08", "V09", "V10"],
    "sign": ["HELLO", "HELLO", "HELP", "HELP", "HELP",
             "YES", "YES", "NO", "NO", "MEDICINE"],
    "signer": ["S01", "S02", "S01", "S03", "S03",
               "S02", "S04", "S01", "S04", "S02"],
    "environment": ["indoor", "indoor", "outdoor", "outdoor", "outdoor",
                    "indoor", "indoor", "outdoor", "indoor", "indoor"],
    "quality": [5, 4, 2, 3, 2, 5, 4, 5, 5, 4]
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
df.groupby(["sign", "environment"]).size()
```

```python
df.groupby(["sign", "signer"]).size()
```

```python
df.groupby("environment")["quality"].mean()
```

```python
df.groupby(["sign", "environment"])["quality"].mean()
```

```python
df["sign"].value_counts().plot(kind="bar")
plt.xlabel("Sign")
plt.ylabel("Number of Observations")
plt.title("Observation Count per Sign Class")
plt.show()
```
