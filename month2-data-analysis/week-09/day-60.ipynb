```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "video_id": ["V01", "V02", "V03", "V04", "V05", "V06", "V07", "V08"],
    "sign": ["HELLO", "HELLO", "HELP", "HELP", "HELP", "YES", "YES", "NO"],
    "environment": ["indoor", "outdoor", "outdoor", "outdoor",
                    "indoor", "indoor", "indoor", "outdoor"],
    "quality": [5, 4, 3, 2, 4, 3, 5, 5]
})
```

```python
# Task 1 — Sign distribution
df["sign"].value_counts().plot(kind="bar")
plt.xlabel("Sign")
plt.ylabel("Number of Videos")
plt.title("Distribution of Videos per Sign")
plt.show()
```

```python
# Task 2 — Quality distribution
df["quality"].plot(kind="hist", bins=5, edgecolor="red")
plt.xlabel("Quality Score")
plt.ylabel("Frequency")
plt.title("Distribution of Video Quality Scores")
plt.show()
```

```python
# Task 3 — Average quality by sign
df.groupby("sign")["quality"].mean().plot(kind="bar")
plt.xlabel("Sign")
plt.ylabel("Average Quality Score")
plt.title("Average Video Quality Score by Sign")
plt.show()
```
